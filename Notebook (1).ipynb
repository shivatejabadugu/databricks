```python
data = [
    (1,'kiran','Hyderabad',30),
    (2,'ravi','Chennai',35),
    (3,'priya','Bangalore',28)
]
columns = ['id','name','city','age']
df = spark.createDataFrame(data,columns)
df.show()          # data print (default 20 rows)
df.printSchema()   # column names + types
df.count()         # entha rows unnayi - 3
df.columns         # column names list - ['id', 'name', 'city', 'age']
```

```python
df.select("name","city","age").show()
```

```python
from pyspark.sql.functions import col
df.select("name","age").filter(col("age")>=30).show()
```

```python
df2 = df.withColumn("age_after_5_years",col("age")+5)
df2.show()
df2.select("name","age_after_5_years").show()
df2.select("name","age_after_5_years").filter(col("age_after_5_years")>=35).show()
df2.select("name","age_after_5_years").filter(col("age_after_5_years")>=35).withColumnRenamed("age_after_5_years","new_age").show()
```
